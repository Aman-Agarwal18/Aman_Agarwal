```python
import os
import streamlit as st
import pickle
import time
import langchain
from langchain_mistralai.chat_models import ChatMistralAI
from langchain import PromptTemplate, LLMChain
from langchain.chains import RetrievalQAWithSourcesChain
from langchain.chains.qa_with_sources.loading import load_qa_with_sources_chain
from langchain.chains.question_answering import load_qa_chain
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.document_loaders import UnstructuredURLLoader
from langchain_mistralai import MistralAIEmbeddings
from langchain.vectorstores import FAISS
from langchain.memory import ConversationBufferMemory
os.environ['MISTRAL_API_KEY'] ="HeLJao9xAYPjoDzK0nd9WPArs5R5sw3i"
```

After installation, you can initialize the Mistral model. You have already set the API key as an environment variable in a previous cell.

```python
llm = ChatMistralAI(
    model = "mistral-large-latest",
    temperature = 0.2,
    max_tokens = 500
)

loaders = UnstructuredURLLoader(urls=[
    "https://www.moneycontrol.com/news/business/markets/taking-stock-selling-continues-for-7th-day-sensex-nifty-down-marginally-13588146.html",
    "https://www.moneycontrol.com/news/business/stocks/indian-oil-corporation-shares-jump-3-03-amid-volume-surge-alpha-article-13587881.html"

])
data = loaders.load()
len(data)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 200
)
docs = text_splitter.split_documents(data)
len(docs)
```

```python
docs[5]
```

```python
%pip install faiss-cpu
```

```python
embedding = MistralAIEmbeddings()
vectorstore = FAISS.from_documents(docs, embedding)
```

```python
file_path = "vector_index_mistral" # Save directory, not a single file
vectorstore.save_local(file_path)
```

```python

index_file = os.path.join(file_path, "index.pkl")
if os.path.exists(index_file):
    with open(index_file, "rb") as f:
        VectorIndex = pickle.load(f)
```

```python
chain = RetrievalQAWithSourcesChain.from_llm(llm = llm, retriever=vectorstore.as_retriever())
chain
```

```python
query = "who are the top losers on the Nifty?"

langchain.debug = True

chain({"question":query}, return_only_outputs=True)
```
